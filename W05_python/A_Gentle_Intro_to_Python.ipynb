### 1.1 What is Python?

Python is a programming language. A programming language is an **artificial language** that humans use to communicate with computers. 

We call it a language because it has its own vocabulary and grammatical rules. It is artificial because it is invented by humans. It is different from the natural languages we speak, such as English and German. Unlike natural languages, which evolve over time through use, programming languages are deliberately designed. Their rules are also much stricter, so even a small typo can stop your code from running.

### 2.1 Variables

We can assign a value to a variable using the equals sign `=`. Variable names can include letters, numbers, and underscores. Variable names are case sensitive. Variable names can’t start with a number: `trip_1` is valid, but `1_trip` isn’t.

````python
# btw, text after a hashtag (#) is a comment. Python ignores it when running the code.
num_buses = 5 # an integer (i.e., a whole number)
speed_kmh = 23.4  # a float (i.e., a number that has a decimal point)
travel_mode = "tram"  # a string (i.e., a sequence of characters)
````

````python
# Check data type
print(type(num_buses))
print(type(speed_kmh))
print(type(travel_mode))
````

````python
# We can add an integer and a float
# Python runs this even though adding buses and speed has no real-world meaning
print(num_buses + speed_kmh)
````

````python
# This cell gives an error on purpose: we can't add an integer and a string
print(num_buses + travel_mode)
````

````python
# We can't do math with strings, but we can concatenate (i.e., join) them
# To join a number and a string, first convert the number to a string
print(str(num_buses) + " buses")
````

````python
# We can use commas in print() to combine values and text
# This is useful for debugging and describing results
print(num_buses, "should be a whole number")
````

### 2.2 Lists

A list is one of the most common data structures in Python. It stores a series of items **in a particular order** and those items don't all need to be the same type. We access an item using its **index**, which indicates its position in the list. Python starts counting at zero, so the first item has index 0, the second has index 1, and so on. Below are some basics. To learn more, see this [here](https://www.analyticsvidhya.com/blog/2021/06/15-functions-you-should-know-to-master-lists-in-python/).

````python
# Create an empty list with square brackets
empty_list = []
````

````python
# Create a list by separating items with commas
trip_info = ["tram", 12.5, 3]  # travel mode, distance in km, number of stops
print(trip_info)
````

````python
# Access an item using its index
print(trip_info[0])
````

````python
# Add an item to the end of a list
trip_info.append("on time")
print(trip_info)
````

````python
# Remove an item from a list by its index
trip_info.pop(-1)
print(trip_info)
````

### 2.3 Loops

Loops are essential in Python.

<img src="https://files.realpython.com/media/Python-for-Loops-Definite-Iteration_Watermarked.b38126d495e1.jpg" width=50% align="left">
<img src="https://files.realpython.com/media/Python-while-Loops-Indefinite-Iteration_Watermarked.ef0b89ed9dae.jpg" width=50% align="right">


Loops let us repeat a block of code without rewriting it. Python has two main types of loops:

- A **`for` loop** runs once for each item in a sequence, such as a list.
- A **`while` loop** runs for as long as a condition is true.

The indented code below the `for` or `while` line is the block that gets repeated. Python uses this indentation to determine which lines belong to the loop.

````python
# Loop through all the items in trip_info
for item in trip_info:
    print(item)
````

The `while` loop below uses a counter to keep track of its position in the list. The function `len()` returns the number of items in a list.

````python
# Check the length of trip_info
print(len(trip_info))
````

````python
counter = 0  # while loops are often accompanied by a counter

# Keep looping while counter is less than the length of trip_info
while counter < len(trip_info):
    print("counter =", counter, "item =", trip_info[counter])
    counter += 1  # shorthand for counter = counter + 1
````

Each pass through the loop increases the counter by 1. The loop stops when `counter` equals the length of `trip_info` because the condition is no longer true.

Without `counter += 1`, the condition would stay true and the loop would run forever. This is called an **infinite loop**. If this happens, stop the cell by selecting **Kernel > Interrupt** in the menu.

### 2.4 Conditions

`if` statements let our code examine logical conditions and act based on different conditions. Python checks each condition in order, runs the block for the first one that is true, and skips the rest. If none are true, it runs the `else` block.

&#x270A; **<font color=firebrick>DO THIS: </font> Run the cell below. Then change `distance_km` and `has_bike` to get each of the four possible results.**

````python
distance_km = 3
has_bike = True # change True to False

if distance_km <= 5 and has_bike:  # both conditions must be true
    print("Ride your bike.")
elif distance_km <= 2:  # checked only if the condition above is false
    print("Walk.")
elif distance_km <= 20:  # checked only if the conditions above are false
    print("Take the tram.")
else:  # runs if none of the conditions above are true
    print("Take the train.")
````

### 2.5 Functions

A function is a reusable block of code designed to do one specific job. Instead of rewriting the same code, we can call the function whenever we need to do that task.

A variable created inside a function is a **local variable**, which means it exists only inside that function. In other words, what happens in a function stays in a function! To use a value outside the function, we need to send it back with `return`.

````python
def miles_to_km(miles):
    """Convert miles to kilometers."""
    kilometers = miles * 1.609344  # 1 mile is equal to 1.609344 km
    return kilometers
````

````python
# View the function's docstring
help(miles_to_km)
````

````python
print("5 miles is equal to", miles_to_km(5), "kilometers.")
````

&#x2753;**Question: Look at `miles_to_km` above. What parts make up a function?**

<details>
<summary>Click to see the answer</summary>

The general syntax is:

```python
def function_name(parameters):
    """Docstring"""
    statements
    return outputs
```

Notice that:

* The keyword `def` marks the start of a function definition.
* The function name is what we use to call the function. It should describe what the function does.
* Parameters are the input names listed in the function definition, such as `miles`. Arguments are the values we pass when calling the function, such as `5`. Not all functions need inputs.
* A colon (`:`) must follow the closing parenthesis.
* A docstring, short for documentation string, describes what the function does. It is optional but recommended.
* One or more indented statements make up the function body. The standard indentation is 4 spaces.
* `return` ends the function and sends one or more values back to the code that called it. Without a `return` statement, the function returns `None`, Python's way of representing no value. It's good practice to use `return` when you need to use the result later.

</details>

To call a function, write its name followed by parentheses, including any required arguments:

````python
miles_to_km(10)
````

## Agenda 3. A mapping demo


So far, we've learned what Python is and covered the basics. Now let's apply them to a real dataset.

Transportation planners often want to know who uses their services and what those users think. Travel surveys are a common way to find out, but they are expensive and time-consuming. Social media offers another, often cheaper, source of information, since people share their experiences and complain directly to transport operators.

Today, we'll use Twitter (now X) data to explore where people who tweet about Darmstadt's public transport say they are from.

### 3.1 Load the data

The dataset (`demo_data.csv`) contains location-enabled tweets directed at RMV, HEAG mobilo, and Deutsche Bahn (@RMVdialog, @HEAGmobilo, or @DB_Bahn) from 2012 to 2023. The tweets were sent within a 5-mile radius of Darmstadt city center (49.87167, 8.65027). Retweets are excluded.

Make sure `demo_data.csv` is in the same folder as this notebook. We'll use the `read_csv()` function from pandas to load it as a **dataframe**, which is a table with rows and columns.

````python
df = pd.read_csv("demo_data.csv")
````

````python
df.head()  # show the first few rows
````

### 3.2 Prepare the data

As you can see above, each row in the dataset is a tweet, but we want to map users. We'll keep only the columns we need and remove repeated rows so that each user's location appears only once. We'll also remove rows with missing information because we can't map a user without coordinates.

````python
users = df[["user_id", "user_location", "user_lat", "user_lon"]]  # select these columns
users = users.drop_duplicates()  # remove repeated rows
users = users.dropna()  # remove rows with missing values

print(len(users)) # you should see 166
````

### 3.3 Create the map

First, we create a map centered on Darmstadt. Then, we use a `for` loop to add one marker for each row. Here, `range(len(users))` produces the numbers from 0 to 165, and `.iloc[i]` selects the value at position `i`, much like using an index with a list.

If the map doesn't appear, select **File > Trust Notebook** and run the cell again.

````python
user_map = folium.Map(location=[49.87167, 8.65027], zoom_start=6) # create the map

for i in range(len(users)):
    folium.Marker(
        location=[users.user_lat.iloc[i], users.user_lon.iloc[i]],
        popup=users.user_location.iloc[i],
    ).add_to(user_map)

user_map
````
